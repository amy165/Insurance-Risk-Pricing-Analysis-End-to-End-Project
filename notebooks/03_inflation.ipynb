```python
import pandas as pd

inflation = pd.DataFrame({
    "year": [2011, 2012, 2013, 2014, 2015, 2016, 2017, 2018],
    "inflation_rate": [0.3325, 0.2360, 0.0746, 0.0689, 0.0957, 0.0663, 0.1069, 0.1383]
})

base_year = 2017
```

```python
# Crear índice CPI relativo con 2017 = 100
inflation["factor_to_2017_base"] = None
inflation.loc[inflation["year"] == base_year, "factor_to_2017_base"] = 1
```

```python
CF=1
for y in range(base_year - 1, 2010, -1):
    IR=inflation.loc[inflation["year"] == y+1, "inflation_rate"].values[0] #inflation rate
    CF=CF*(1+IR) #cumulative factor
    inflation.loc[inflation["year"] == y, "factor_to_2017_base"] = CF

IR=inflation.loc[inflation["year"] == 2018, "inflation_rate"].values[0] #inflation rate
inflation.loc[inflation["year"] == 2018, "factor_to_2017_base"] = 1/(1+IR)

```
